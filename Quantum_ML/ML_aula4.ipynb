### O código abaixo inicializa um circuito quântico de $n$ qubits com um estado que é combinação linear dos estados com peso de Hamming igual a $k$ 

```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from math import sqrt
import itertools
```

```python
def create_W_state(n, k):

    # Lista de todos os estados da base em binário com n bits
    basis_states = ["".join(seq) for seq in itertools.product("01", repeat=n)]
    
    # Selecionar apenas os estados com peso de Hamming igual a k
    valid_states = [state for state in basis_states if state.count("1") == k]
    
    # Número total de estados válidos e amplitude uniforme
    num_valid_states = len(valid_states)
    amplitude = 1 / sqrt(num_valid_states)
    
    # Criar o vetor de estado com amplitudes para os estados válidos
    state_vector = [0] * (2**n)
    for state in valid_states:
        position = int(state, 2)  # Converter o estado binário para decimal
        state_vector[position] = amplitude  # Atribuir a amplitude uniforme
    
    # Inicializar o circuito com o vetor de estado
    circuit = QuantumCircuit(n)
    circuit.initialize(state_vector, range(n))
    
    return circuit
```

```python
# Configurações do estado
n_qubits = 4
hamming = 3

# Cria o circuito com a condição inicial desejada
circuit = create_W_state(n_qubits, hamming)
```
