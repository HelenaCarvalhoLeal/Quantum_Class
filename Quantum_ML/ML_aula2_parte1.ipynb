```python
import numpy as np
import pandas as pd
#erro na linha 4 nao sei pq
from sklearn.model_selection import train_test_split

import matplotlib.pyplot as plt
```

```python
class LinearRegression():

  def __init__(self, learning_rate, iterations):

    self.learning_rate = learning_rate
    self.iterations = iterations


  # Onde o aprendizado de maquina acontece
  # Função de treino

  def fit(self, X, Y):

    # N: tamanho do nosso dataset (número de observações)
    # K: número de variáveis


    self.N, self.K = X.shape

    # inicializamos nosso modelo/hipótese

    self.m = np.zeros(self.K)

    self.b = 0

    self.X = X
    self.Y = Y

    # gradient descent

    for i in range(self.iterations):
      self.update_weights()

    return self

  def update_weights(self):

    # predição

    Y_pred = self.predict( self.X )

    # gradientes

    dm = (self.X.T.dot(Y_pred - self.Y ))/self.N

    db = np.sum(Y_pred - self.Y)/self.N

    # evolução dos pesos

    self.m = self.m - self.learning_rate * dm
    self.b = self.b - self.learning_rate * db

    return self

  # inferência

  def predict(self, X):

    # y_pred = mx + b
    return X.dot(self.m) + self.b
```

```python
def main():

  df = pd.read_csv("salary_data.csv")

  X = df.iloc[:, :-1].values
  Y = df.iloc[:, 1].values

  # Divisão dos dados entre conjunto de treino e conjunto de teste

  X_train, X_test, Y_train, Y_test = train_test_split(
    X, Y, test_size = 1/3, random_state = 0 )


    # Treinamento
  model = LinearRegression( iterations = 100, learning_rate = 0.001 )

  model.fit(X_train, Y_train)


  # inferência

  Y_pred = model.predict(X_test)

  for i in range(3):
    print( "Valores preditos:\t{} anos de exp. -> R$ {}".format(X_test[i][0], np.round( model.predict(X_test[i]), 2 )) )

    print( "Valores reais: R$", Y_test[i], "\n" )

  print( "m final          ", round( model.m[0], 2 ) )

  print( "b final          ", round( model.b, 2 ) )

  # Visualização do conjunto de teste

  plt.scatter( X_test, Y_test, color = 'blue' )

  plt.plot( X_test, Y_pred, color = 'orange' )

  plt.title( 'Salário vs Experiência' )

  plt.xlabel( 'Anos de experiência' )

  plt.ylabel( 'Salário' )

  plt.show()

if __name__ == "__main__" :

    main()
```
